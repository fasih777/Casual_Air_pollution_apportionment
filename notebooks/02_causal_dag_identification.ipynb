# Causal DAG Identification
This notebook visualizes the Causal Directed Acyclic Graph (DAG) for our Air Apportionment project using DoWhy.

```python
import sys
import os
import pandas as pd
import numpy as np

# Add src to path
sys.path.append(os.path.abspath('../'))

from src.models.causal_dag import AirApportionmentDAG
```

```python
# Instantiate the DAG
dag = AirApportionmentDAG()

# Create some dummy data to initialize the DoWhy model
np.random.seed(42)
n = 100
df = pd.DataFrame({
    'temp': np.random.normal(25, 5, n),
    'humidity': np.random.normal(50, 10, n),
    'wind_u': np.random.normal(0, 2, n),
    'wind_v': np.random.normal(0, 2, n),
    'pblh': np.random.normal(1000, 200, n),
    'proxy_traffic': np.random.normal(5, 1, n),
    'proxy_industry': np.random.normal(3, 1, n),
    'proxy_dust': np.random.normal(10, 3, n),
    'proxy_biomass': np.random.normal(2, 0.5, n),
    'pm25': np.random.normal(50, 15, n)
})

# Create DoWhy model for one treatment to visualize the graph
model = dag.create_model(df, treatment='proxy_traffic')

# View the model graph
model.view_model()
from IPython.display import Image, display
display(Image(filename="causal_model.png"))
```
